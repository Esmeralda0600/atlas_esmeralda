```python
import xarray as xr
```

```python
data = xr.open_dataset("../data/raw/ToSdW/2022/ERA5_mexico_2022.nc")
data
```

```python
data = data.rename({
    "latitude": "lat",
    "longitude": "lon"
})
#mex_era
```

```python
mex_era = xr.Dataset()
```

```python
mex_era["t2m"] = data["t2m"] - 273.15
mex_era["t2m"].attrs["units"] = "°C"
```

```python
mex_era["ssrd"] = data["ssrd"] / 3600
mex_era["ssrd"].attrs["units"] = "W/m²"
```

```python
mex_era = mex_era.chunk({
    "time": 168, 
    "lat": -1, 
    "lon": -1
})
```

```python
mex_era.to_zarr("../zarr/era5_mexico.zarr", mode="w")
```
